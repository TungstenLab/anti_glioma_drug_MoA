# Visualize overlap between DEGs

```r
# Install packages
install.packages(c("readxl", "ggVennDiagram", "ggplot2", "dplyr", "writexl"))

# Load libraries
library(readxl)
library(ggVennDiagram)
library(ggplot2)
library(dplyr)
library(writexl)
```

```r
# Define file paths
auf_up <- "../../analysis/DEGs_AUF_upregulated.xlsx"
cis_up <- "../../analysis/DEGs_CIS_upregulated.xlsx"
trans_up <- "../../analysis/DEGs_TRANS_upregulated.xlsx"

# Read Excel files
deg1 <- read_excel(auf_up)
deg2 <- read_excel(cis_up)
deg3 <- read_excel(trans_up)

# Column containing the genes
gene_col <- "gene"

genes_auf <- unique(na.omit(as.character(deg1[[gene_col]])))
genes_cis <- unique(na.omit(as.character(deg2[[gene_col]])))
genes_trans <- unique(na.omit(as.character(deg3[[gene_col]])))

# Organize gene sets
gene_sets <- list(
  "AUF" = genes_auf,
  "CIS" = genes_cis,
  "TRANS" = genes_trans
)

venn_plot <- ggVennDiagram(
  gene_sets,
  label = "count",
  label_alpha = 0,
  edge_size = 0.8,
  set_size = 4
) +
  scale_fill_gradient(
    low = "#E8F1FA",
    high = "#2563A6"
  ) +
  theme(
    legend.position = "none",
    text = element_text(family = "sans", size = 12)
  )

# Display diagram
print(venn_plot)

# Save plot as PDF
ggsave(
  filename = "DEGs_up_Venn.pdf",
  plot = venn_plot,
  width = 7,
  height = 6)
```
